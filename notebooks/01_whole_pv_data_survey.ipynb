```python
import duckdb as my_duck
import pandas as pd
import matplotlib.pyplot as plt
```

```python
con = my_duck.connect()
pv_data = con.execute("""
    SELECT *
    FROM read_parquet('../data/silver/silver_pv.parquet')
""").df()
```

## Detecting Missing data - NaN, nulls, etc

```python
nan_by_region = (
    pv_data.groupby("region_id")[
        ["generation_mw", "capacity_mwp", "installedcapacity_mwp", "load_factor"]
    ]
    .apply(lambda x: x.isna().sum())
)

null_by_region = (
    pv_data.groupby("region_id")[
        ["generation_mw", "capacity_mwp", "installedcapacity_mwp", "load_factor"]
    ]
    .apply(lambda x: x.isnull().sum())
)

display(nan_by_region)
display(null_by_region)
```

```python
time_diff_by_region = (
    pv_data
    .sort_values(["region_id", "time"])
    .groupby("region_id")["time"]
    .diff()
    .groupby(pv_data["region_id"])
    .value_counts()
)

display(time_diff_by_region)
```
